```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Setting plot style
sns.set_style("whitegrid")
plt.rcParams['figure.dpi'] = 100
```

```python
# Load both datasets
df_math = pd.read_csv('../data/student-mat.csv', sep=';')
df_por = pd.read_csv('../data/student-por.csv', sep=';')
```

```python
# Basic info
print("=== MATH DATASET ===")
print(df_math.info())
print("\n", df_math.describe())
print("\n", df_math.head())

print("\n=== PORTUGUESE DATASET ===")
print(df_por.info())
print("\n", df_por.describe())
print("\n", df_por.head())
```

```python
# Check shapes
print(f"\nMath dataset: {df_math.shape[0]} rows, {df_math.shape[1]} columns")
print(f"Portuguese dataset: {df_por.shape[0]} rows, {df_por.shape[1]} columns")
print(f"Combined (if we merge): {df_math.shape[0] + df_por.shape[0]} rows")
```

```python
# 1. Check for missing values
print("Missing values in Math dataset:")
print(df_math.isnull().sum())
print("\nMissing values in Portuguese dataset:")
print(df_por.isnull().sum())
```

```python
# 2. Look at the target variable (G3 - final grade)
print("\n=== TARGET VARIABLE: G3 (Final Grade) ===")
print("Math G3 distribution:")
print(df_math['G3'].describe())
print("\nPortuguese G3 distribution:")
print(df_por['G3'].describe())
```

```python
# 3. Check data types
print("\n=== DATA TYPES ===")
print(df_math.dtypes)
```

```python
# 4. See unique values in categorical columns (sample a few)
print("\n=== SAMPLE CATEGORICAL VALUES ===")
print("Schools:", df_math['school'].unique())
print("Study time levels:", df_math['studytime'].unique())
print("Failures:", df_math['failures'].unique())
```
