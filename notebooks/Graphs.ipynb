```python
import pandas as pd
```

```python
# Load both datasets
df_math = pd.read_csv('../data/student-mat.csv', sep=';')
df_por = pd.read_csv('../data/student-por.csv', sep=';')
```

```python
# Basic info
print("=== MATH DATASET ===")
print(df_math.info())
print("\n", df_math.describe())
print("\n", df_math.head())

print("\n=== PORTUGUESE DATASET ===")
print(df_por.info())
print("\n", df_por.describe())
print("\n", df_por.head())
```

## Graphs for Math Subject

### Histogram Graph of G3

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = df_math
sns.set(style="whitegrid")

# 1) Histogram of G3 
plt.figure(figsize=(6,4))
plt.hist(df["G3"], bins=10, edgecolor="black")
plt.title("Distribution of Final Grade (G3)")
plt.xlabel("G3")
plt.ylabel("Count")
plt.show()
```

### Bar Chart Graph

```python
# 2) Bar chart: studytime distribution
plt.figure(figsize=(6,4))
study_counts = df["studytime"].value_counts().sort_index()
plt.bar(study_counts.index.astype(str), study_counts.values, edgecolor="black")
plt.title("Study Time Distribution")
plt.xlabel("Study time category (studytime)")
plt.ylabel("Number of students")
plt.show()
```

### Box Plot Graph

```python
# 3) Box plot: grades by school
plt.figure(figsize=(6,4))
sns.boxplot(data=df, x="school", y="G3")
plt.title("Final Grade (G3) by School")
plt.xlabel("School")
plt.ylabel("G3")
plt.show()
```

### Scatter Plot Graph

```python
# 4) Scatter: absences vs grades
plt.figure(figsize=(6,4))
plt.scatter(df["absences"], df["G3"], alpha=0.6)
plt.title("Absences vs Final Grade (G3)")
plt.xlabel("Absences")
plt.ylabel("G3")
plt.show()
```
