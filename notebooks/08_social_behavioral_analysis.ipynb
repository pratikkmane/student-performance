```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Data

```python
df = pd.read_csv("../data/processed/combined_student_data.csv")
```

```python
df.shape
df.columns
```

### 1. Weekday Alcohol (Dalc)

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='Dalc', y='G3', data=df)
plt.title("Weekday Alcohol Consumption vs Final Grade")
plt.xlabel("Weekday Alcohol Consumption (1=Low, 5=High)")
plt.ylabel("Final Grade (G3)")
plt.show()
```

### Weekday Alcohol vs Final Grade

The visualization suggests a negative relationship between weekday alcohol consumption and final grades. Students with lower alcohol consumption levels tend to have higher median grades, while higher alcohol levels are associated with lower academic performance.

### 2. Weekend Alcohol (Walc)

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='Walc', y='G3', data=df)
plt.title("Weekend Alcohol Consumption vs Final Grade")
plt.xlabel("Weekend Alcohol Consumption (1=Low, 5=High)")
plt.ylabel("Final Grade (G3)")
plt.show()
```

### Weekend Alcohol vs Final Grade

The boxplot indicates a slight negative association between weekend alcohol consumption and final grades. Higher weekend alcohol levels appear to correspond with lower median academic performance.

### 3. Going Out Frequency (goout)

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='goout', y='G3', data=df)
plt.title("Going Out Frequency vs Final Grade")
plt.xlabel("Going Out Frequency (1=Low, 5=High)")
plt.ylabel("Final Grade (G3)")
plt.show()
```

### Going Out Frequency vs Final Grade

The visualization suggests that students who go out more frequently tend to have slightly lower median grades. Higher social activity levels may be associated with reduced academic performance.

### 4. Health Status (health)

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='health', y='G3', data=df)
plt.title("Health Status vs Final Grade")
plt.xlabel("Health Status (1=Poor, 5=Excellent)")
plt.ylabel("Final Grade (G3)")
plt.show()
```

### Health Status vs Final Grade

The relationship between health status and final grades appears weak. There is no strong or consistent pattern indicating that health level significantly impacts academic performance.

### 5. Absences vs Final Grade

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='absences', y='G3', data=df)
plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")
plt.show()
```

### Correlation Values

```python
df[['absences','G3']].corr()
```

### Absences vs Final Grade

The scatter plot shows a negative relationship between absences and final grades. Students with higher numbers of absences generally tend to have lower academic performance. Absences appear to be one of the stronger behavioral predictors of final grade.

### 6. Romantic Relationship vs Final Grade

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='romantic', y='G3', data=df)
plt.title("Romantic Relationship vs Final Grade")
plt.xlabel("In a Romantic Relationship (yes/no)")
plt.ylabel("Final Grade (G3)")
plt.show()
```

### Romantic Relationship vs Final Grade

The visualization suggests a slight difference in academic performance between students in romantic relationships and those who are not. However, the overall impact appears moderate compared to other behavioral factors such as absences or alcohol consumption.

## Key Findings – Social & Behavioral Factors

- Higher alcohol consumption (weekday and weekend) is associated with lower median grades.
- Increased frequency of going out shows a slight negative relationship with academic performance.
- Absences show a clear negative relationship with final grade and appear to be one of the strongest behavioral predictors.
- Health status shows minimal impact on academic performance.
- Romantic relationships show only a moderate effect on grades.

Overall, absences and high alcohol consumption appear to be the most significant behavioral risk factors affecting student performance.
