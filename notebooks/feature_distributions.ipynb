Load Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Set style
sns.set(style="whitegrid")

# Load combined dataset
df = pd.read_csv("../data/processed/combined_student_data.csv") #will upload the combined dataset after Emmanuel work on it

# Quick check
df.head()
```

```python
df.shape
df.info()
```

Basic Dataset Overview

1. Study Time

A. Histogram

```python
plt.figure(figsize=(6,4))
sns.histplot(df['studytime'], bins=4, kde=False)
plt.title("Distribution of Study Time")
plt.xlabel("Study Time (1=low, 4=high)")
plt.ylabel("Count")
plt.show()
```

B. Boxplot vs Final Grade

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='studytime', y='G3', data=df)
plt.title("Final Grade vs Study Time")
plt.show()
```

Most students study at level 2. Higher study time levels show slightly higher median final grades, indicating a positive relationship between study effort and performance.

2. Past Failures

```python
plt.figure(figsize=(6,4))
sns.histplot(df['failures'], bins=4)
plt.title("Distribution of Past Failures")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='failures', y='G3', data=df)
plt.title("Final Grade vs Past Failures")
plt.show()
```

Students with more past failures show significantly lower median final grades. This appears to be a strong predictive feature.

3. Mother’s Education

```python
plt.figure(figsize=(6,4))
sns.histplot(df['Medu'], bins=5)
plt.title("Distribution of Mother's Education")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='Medu', y='G3', data=df)
plt.title("Final Grade vs Mother's Education")
plt.show()
```

Students whose mothers have higher education levels tend to show slightly improved academic outcomes.

4. Absences

```python
plt.figure(figsize=(6,4))
sns.histplot(df['absences'], bins=20)
plt.title("Distribution of Absences")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='risk_category', y='absences', data=df)
plt.title("Absences by Risk Category")
plt.show()
```

Absences are right-skewed. Students in the high-risk category generally have more absences.

5. Age

```python
plt.figure(figsize=(6,4))
sns.histplot(df['age'], bins=8)
plt.title("Distribution of Age")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='age', y='G3', data=df)
plt.title("Final Grade vs Age")
plt.show()
```

Most students are between 15–18 years old. Age does not appear to strongly impact final grades.

6. Key Findings:


Past failures show strong negative correlation with final grade.

Study time shows moderate positive relationship.

Absences may influence risk category.

Parent education has mild impact.

Age likely weak predictor.
